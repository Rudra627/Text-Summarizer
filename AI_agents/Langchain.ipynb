Create tools

Tools let a model interact with external systems by calling functions you define. Tools can depend on runtime context and also interact with agent memory.

```python
@tool
def get_whether(location: str) -> str:
  """Returns the temperature for a given location."""
  return f"temperature of {location} is 100C"
```

```python
@tool
def get_time(location:str)->str:
  """Return the Current time of the Location"""
  time=model1.invoke(f"Tell me the Time of this {location} in number only")
  return time.content
```

```python
agent=create_agent(
    model=model1,
    tools=[get_whether,get_time],
    system_prompt="Hello You are Good AI Agents Your Goal is To Acess the Tool properly"
)
```

Pydantic


**Structured output**

Return a validated schema from the agent using response_format=. See Structured output for strategies and examples.

```python
from pydantic import BaseModel,Field
class Movie(BaseModel):
  title:str=Field(description="The Movie Title is:")
  year:int=Field(description="The Movie relese in the Year")
  theme:str=Field(description="The Movie is Based on theme of")
  rating:float=Field(description="Movie has a rating of")
```

```python
model_with_structure=model1.with_structured_output(Movie)
res=model_with_structure.invoke("Tell me About the Movie The Godfather")
print(res)
```

```python
res2=model.invoke("Tell me About the Movie The Godfather")
print(res2.content)
```

Short-term memory

```python
from langgraph.checkpoint.memory import InMemorySaver
```

```python
agent = create_agent(
    model=model1,
    checkpointer=InMemorySaver(),
)
thread_config = {"configurable": {"thread_id": "1"}}
response = agent.invoke(
    {"messages": [{"role": "user", "content": "Hi! My name is Bob."}]},
    thread_config,
)["messages"][-1].content
print(response)
```

```python
response = agent.invoke(
    {"messages": [{"role": "user", "content": "What's my name?"}]},
    thread_config,
)["messages"][-1].content

print(response)
```
